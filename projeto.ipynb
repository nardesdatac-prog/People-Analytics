## People Analytics: Employee Insights Dashboard

## Motivo do projeto -
    Desenvolver uma solução de análise de dados voltada para Recursos Humanos, utilizando indicadores de colaboradores para transformar informações brutas em insights estratégicos sobre retenção, distribuição da força de trabalho, remuneração e perfil dos funcionários.

Exploração e preparação dos dados -

```python
#Importando a biblioteca e carregando o dataset
import pandas as pd

df = pd.read_csv("../WA_Fn-UseC_-HR-Employee-Attrition.csv")
df.head()
```

```python
df.describe(include='object')
```

## Perguntas de negócio -

### 1. Total de funcionários ativos:

```python
# Filtrando os funcionários ativos
df_ativos = df[df["Attrition"] == "No"].copy()
df_ativos["Attrition"].value_counts()
```

Total de 1233 funcionários ativos.

### 2. Porcentagem de Turnover:

```python
turnover = (df["Attrition"].value_counts(normalize=True)["Yes"] * 100).round(2)

turnover
```

O total de 16,12% de Turnover

O cálculo do turnover permite medir a proporção de funcionários que deixaram a empresa, sendo um indicador importante para avaliar retenção de talentos e possíveis impactos na gestão de pessoas.

## 3) Idade média para funcionários ativos:

```python
df_ativos["Age"].mean().round(2)
```

A média para a idade de funcionários é de 37 anos.

## 4) Salário médio:

```python
df_ativos["MonthlyIncome"].mean().round(2)
```

O salário médio dos funcionários é de 6.832,74.

## 5) Total de funcionários por departamento:

```python
df_ativos["Department"].value_counts()
```

Pesquisa e Desenvolvimento: 828.
Vendas: 354.
RH: 51.

## 6. Percentual de funcionários por gênero:

```python
df_ativos["Gender"].value_counts(normalize=True).mul(100).round(2)
```

Homens: 59,37%
Mulheres: 40,63%

## 7) Tempo médio de empresa por departamento:

```python
df_ativos.groupby("Department")["YearsAtCompany"].mean().round(2)
```

RH & Vendas possuem funcionários com quase 8 anos de empresa, enquanto Pesquisa e Desenvolvimento posue funcionários com 7 anos. 

## 8. Total de funcionários por cargo:

```python
df_ativos["JobRole"].value_counts().sort_values(ascending=False)
```

A distribuição dos funcionários por cargo permite identificar quais funções concentram a maior parte da força de trabalho, auxiliando no planejamento de equipes, dimensionamento de recursos e análise da estrutura organizacional.

## 9) Faixa salarial por cargo:

```python
df_ativos.groupby("JobRole")["MonthlyIncome"].mean().round(2).sort_values(ascending=False)
```

A análise salarial por cargo permite compreender a distribuição da remuneração dentro da empresa, identificando diferenças entre funções e auxiliando em decisões relacionadas à estrutura salarial e planejamento de carreira.

## Visualização de Dados:
    Para criar um dashboard interativo no Power BI, será feito antes a separação de colunas necessárias para que as análises feitas aqui estejam expostas da maneira correta, eliminando qualquer outra que não faça sentido com o projeto.

```python
colunas_dashboard = [
    "Attrition",
    "Department",
    "Gender",
    "Age",
    "MonthlyIncome",
    "JobRole",
    "YearsAtCompany"
]

df_dashboard = df[colunas_dashboard].copy()
```
